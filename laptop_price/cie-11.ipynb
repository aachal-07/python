```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
def import_to_array(file_path):
    if file_path():
        df= pd.read(file_path)
        return df.values  
```

```python
def basic_statistics(arr):
    total_sum = np.sum(arr)
    mean_val = np.mean(arr)
    var_val = np.var(arr)
    std_val = np.std(arr)
    return total_sum, mean_val, var_val, std_val
```

```python
def plot_array(arr, title="Array Plot"):
    plt.figure()
    plt.plot(arr.flatten())
    plt.title(title)
    plt.xlabel("Index")
    plt.ylabel("Value")
    plt.show()
```
