EXCEPTION HANDLING IN OOPS

```python
try:
    pass  # main code
except:
    pass # exception code and handling
else:
    pass # executed when there is no exception 
finally: 
    pass  ### always executed
```

```python
try:
    a = 10 
    b = 5
    print(a/b)
except:
    print("exception")
```

```python
try: 
    num1 =int(input("enter a number"))
    num2 =int(input("enter a second number"))
    print(num1/num2)
except:
    print("am aachal")
```

```python
try: 
    num1 =int(input("enter a number"))
    num2 =int(input("enter a second number"))
    print(num1/num2)
except Exception as e:
    print(e)
    print("am aachal")
```

```python
try: 
    num1 =int(input("enter a number"))
    num2 =int(input("enter a second number"))
    print(num1/num2)
except Exception as e:
    print(e)
    print("am aachal")
else:
    print("am aachal")
finally:
    print("am aachal bagade")
```

```python
# RuntimeError
try:
    x = -6

    if x<0:
        raise 

except:
    print("not valid")
else:
    print("am aachal ")
finally:
    print("am aachal bagade")
```

```python
# valueerror 
try:
    x = -6
    if x<0:
        raise ValueError("not valid") 
except ValueError as e:
    print(e)
```
